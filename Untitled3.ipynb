```python
import sqlite3
from datetime import datetime

DB_FILE = "customer_loans.db"  # your database

# Connect to DB
conn = sqlite3.connect(DB_FILE)
c = conn.cursor()

# Generate next loan_number
c.execute("SELECT loan_number FROM loan_applications ORDER BY id DESC LIMIT 1")
last_loan = c.fetchone()
if last_loan:
    last_number = int(last_loan[0][5:])  # remove 'TRBLN' prefix
    new_number = last_number + 1
else:
    new_number = 1
loan_number = f"TRBLN{new_number:08d}"

# Existing account number
account_number = "TRB50300002"

# Insert new loan
c.execute("""
INSERT INTO loan_applications
(account_number, loan_number, loan_type, amount, tenure, purpose, application_date, status)
VALUES (?, ?, ?, ?, ?, ?, ?, ?)
""", (
    account_number,
    loan_number,
    "education",
    200000,
    48,  # assuming 48 months for B.E
    "B.E",
    "2025-10-05 12:00:00",  # applied date
    "Need to visit branch on 2025-10-10"
))

conn.commit()
conn.close()

print(f"Inserted new loan: {loan_number}")
```
